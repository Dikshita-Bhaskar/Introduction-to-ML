## 11. Flatten vs Ravel

_`.flatten()` → returns a copy (original safe) | `.ravel()` → returns a view (original can change too)_


**Q11.1** Create a 2D array, flatten it with both methods, modify the first element of each result, and check which one affects the original array.

```python
# Write your code here
a = np.arange(4).reshape(2,2)
b = a.flatten()
b[0] = 12
print(b)
print(a)
c = a.ravel()
c[0] = 15
print(c)
print(a)
```

## 12. Stacking & Concatenation

_`np.vstack()` → stack rows | `np.hstack()` → stack columns | `np.concatenate(axis=)` → general version_


**Q12.1** Create two 2x2 arrays. Combine them using vstack, hstack, and concatenate on both axis=0 and axis=1.

```python
# Write your code here
a = np.arange(4).reshape(2,2)
b = np.arange(3,7).reshape(2,2)
print(np.vstack((a,b)))
print(np.hstack((a,b)))
print(np.concatenate((a,b),axis=0))
print(np.concatenate((a,b),axis=1))
```
