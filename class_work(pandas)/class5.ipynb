```python
import pandas as pd
import numpy as np

df = pd.read_csv('Titanic-Dataset.csv')              
tip = pd.read_csv('tip.csv')     
```

```python
#aggeration
tip.groupby('day').agg(
    Bill_count=('total_bill', 'size'),
    avg_bill=('total_bill', 'mean'),
    avg_tip=('tip','mean')
)
```

```python
tip.groupby('time').agg(
    bills=('total_bill', 'mean'),
    avg_bill=('total_bill', 'mean'),
    avg_party=('size','mean')
)
```

```python
#table
summary = tip.groupby('day').agg(
    Bill_count=('total_bill', 'size'),
    avg_bill=('total_bill', 'mean'),
    avg_tip=('tip','mean')
)
summary
```

```python
summary[summary['day']=='Sun']
#it doesnt work because when we do groupby on something it gets converted to index(i.e it becomes a row)
```

```python
#columns in a list
print(summary.columns.tolist())
print(summary.index.to_list())
```

```python
#reset index
flat = summary.reset_index()
print(flat.columns.tolist())
flat.head()
```

```python
#make a new file names report
summary.to_csv('Report.csv')
flat = pd.read_csv('Report.csv')
flat.head()
```

```python
smoker_report = tip.groupby('smoker').agg(
    Bill_count=('total_bill', 'size'),
    avg_bill=('total_bill', 'mean'),
    avg_tip=('tip','mean')
)
smoker_report
```

```python
smoker_report.reset_index().to_csv('Smoker.csv')
```
