## Q8
Compare the mean age with the 50% value and state what it suggests about the passengers' ages.


```python
print(t['Age'].describe())
print('Mean:',t['Age'].mean())
print('50%:',t['Age'].median())
print('The mean is larger than the 50% value, suggesting the age distribution is slightly right-skewed.')
```
